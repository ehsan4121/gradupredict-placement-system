# GraduPredict Model Training Notebook
This notebook loads the project dataset, inspects it, trains the compatible NumPy Logistic Regression model, evaluates it, and saves JSON model parameters.

```python
from pathlib import Path
import json, numpy as np, pandas as pd
ROOT=Path('.')
df=pd.read_csv(ROOT/'data'/'placement_demo.csv')
df.head()
```

```python
df.shape, df.isna().sum(), df['placed'].value_counts()
```

```python
FEATURES=['ssc_percentage', 'hsc_percentage', 'degree_percentage', 'employability_score', 'internships', 'projects', 'communication_rating', 'technical_rating', 'backlogs', 'work_experience', 'training_completed']
X=df[FEATURES].to_numpy(float); y=df['placed'].to_numpy(float)
rng=np.random.default_rng(610); idx=rng.permutation(len(y)); cut=int(.8*len(y)); train,test=idx[:cut],idx[cut:]
mean=X[train].mean(0); scale=X[train].std(0); scale[scale==0]=1
```

```python
xs=(X[train]-mean)/scale; w=np.zeros(X.shape[1]); b=0.0
for _ in range(1800):
    p=1/(1+np.exp(-(xs@w+b))); error=p-y[train]
    w-=.08*(xs.T@error/len(train)); b-=.08*error.mean()
probs=1/(1+np.exp(-(((X[test]-mean)/scale)@w+b))); pred=(probs>=.5)
float((pred==y[test]).mean())
```
